```python
product = "Laptop"
qty = 4
unit_price = 35000
unit_cost = 28000

revenue = qty * unit_price
cost = qty * unit_cost
profit = revenue - cost
margin = profit / revenue * 100

if margin >=20:
    rating = "Healthy Margin"
else:
    rating = "Low Margin"

print(f"Revenue: {revenue}")
print(f"Cost: {cost}")
print(f"Profit: {profit}")
print(f"Margin: {margin}",end="%")
print(f" Rating: {rating}")
```
