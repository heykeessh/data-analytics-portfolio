```python
totCost_by_cat = (
    products.groupby('category')['unit_cost']
        .sum()
        .sort_values(ascending = True)
)
```

```python
mulAggbyCat = (
    products.groupby('category')
    .agg(
        totUCost = ('unit_cost', 'sum'),
        totQty = ('unit_cost', 'sum'),
        minUCost = ('unit_cost', 'min'),
        maxUCost = ('unit_cost', 'max'),
        meanUCost = ('unit_cost', 'mean')
    )
    .reset_index()
)
```
