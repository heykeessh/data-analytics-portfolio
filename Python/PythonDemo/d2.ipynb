```python
sales = [45000, 82000, 120000, 67000, 95000]
target = 75000

# 1. Calculate total and average
total_sales = sum(sales)
average_sales = total_sales / len(sales)

# 2. Classify each value
def classify_sales(amount, target):
    if amount >= target:
        return "Met or exceeded target"
    else:
        return "Below target"

# 3. Count values >= target
sales_above_target = sum(1 for amount in sales if amount >= target)

# Display results
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("\nSales Classification:")

# 4. Use a function to classify each value
for amount in sales:
    classification = classify_sales(amount, target)
    print(f"₱{amount:,} - {classification}")

print("\nNumber of values meeting target:", sales_above_target)
```
